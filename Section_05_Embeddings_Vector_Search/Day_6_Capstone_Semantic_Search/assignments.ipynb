# Day 6 — Capstone Assignments


## Exercise 1 — Ingest a real PDF

Point your `/ingest` endpoint at any PDF you have (a research paper, a manual, a book chapter). Verify `/stats` shows a chunk count that looks reasonable (~1–3 chunks per page).


## Exercise 2 — Craft 5 test queries

Come up with 5 queries about the PDF you ingested — a mix of easy (keyword-matching), medium (paraphrased), and hard (implied). Run each and note the top result. Which queries worked best?


## Exercise 3 — Multi-source filter

Ingest a **second** PDF. Run a query with and without the `source` filter. Confirm the filter narrows results to only that file.


## Stretch — Distance threshold

Modify the `/search` endpoint so that any result with `distance > 1.0` is filtered out. If nothing passes, return `{"results": [], "message": "No relevant results found."}`. Test with an obviously off-topic query.
