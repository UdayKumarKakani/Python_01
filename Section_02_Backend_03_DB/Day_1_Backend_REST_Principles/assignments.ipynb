## 🎁 Bonus — Idempotency

**Problem:** Which HTTP methods are *idempotent*, and why does it matter?

An operation is **idempotent** if calling it once vs. many times has the same effect on the server.

| Method | Idempotent? | Why |
|--------|-------------|-----|
| GET    | ✅ | Reading doesn't change state |
| PUT    | ✅ | Replacing with the same body gives the same result |
| DELETE | ✅ | Deleting an already-deleted resource is a no-op |
| POST   | ❌ | Each call typically creates a *new* resource |
| PATCH  | ⚠️ | Depends on the patch (usually not guaranteed) |

**Why it matters:** clients (and proxies) can safely **retry** idempotent calls on a network error. Retrying a non-idempotent `POST` may create duplicates — use an *idempotency key* in production.


````
# Bonus: store the answer as a dict so it can be auto-checked
idempotent = {
    "GET":    True,
    "PUT":    True,
    "DELETE": True,
    "POST":   False,
    "PATCH":  False,   # usually not guaranteed
}

for method, ok in idempotent.items():
    print(f"{method:6s} -> {'idempotent' if ok else 'NOT idempotent'}")
````
