# Day 6 — Capstone Deliverables


## Deliverable 1 — Deployed URL

A public HTTPS URL where the RAG chatbot's `/docs` responds.


## Deliverable 2 — GitHub repo with green CI

Repo link. CI badge in the README. `main` must be green.


## Deliverable 3 — Langfuse traces

Screenshot or Langfuse public-view link showing 10+ traces from real usage.


## Deliverable 4 — Load-test report

p50 / p95 / error rate over a 20-request run. Which endpoint is your bottleneck?


## Deliverable 5 — Stated SLO

Written in your README: "We commit to p95 latency < X seconds and Y% availability."


## Stretch — Portfolio blog post

400-word post: architecture, cost per query, latency, one hard bug you fixed. Publish on Medium / your site / dev.to.
