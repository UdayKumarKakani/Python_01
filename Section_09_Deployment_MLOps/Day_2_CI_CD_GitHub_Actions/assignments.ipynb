# Day 2 — Assignments


## Exercise 1 — First workflow

Push your Day 1 project to a GitHub repo. Add `.github/workflows/ci.yml` with a single test job that runs `pip install -r requirements.txt`. Confirm it goes green in the Actions tab.


## Exercise 2 — Docker build + push

Extend the workflow to build and push to `ghcr.io`. Pull the image with `docker pull ghcr.io/<your-user>/<repo>:latest` and run it. Verify it works.


## Stretch — PR gating

Add a rule: PRs cannot merge if `test` fails. GitHub → Settings → Branches → Add branch protection rule for `main` → Require status checks → `test`.
