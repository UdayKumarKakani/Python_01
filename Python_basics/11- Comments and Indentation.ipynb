# Python Comments

Comments are lines that exist in computer programs that are ignored by compilers and interpreters.

Including comments in programs makes code more readable for humans as it provides some information or explanation about what each part of a program is doing.

In general, it is a good idea to write comments while you are writing or updating a program as it is easy to forget your thought process later on, and comments written later may be less useful in the long term.

In Python, we use the hash (#) symbol to start writing a comment.

````python
# Print Hello, world to console

print("Hello, world")
````

# Multi Line Comments

If we have comments that extend multiple lines, one way of doing it is to use hash (#) in the beginning of each line.

````python
#This is a long comment
#and it extends
#Multiple lines
````

Another way of doing this is to use triple quotes, either ''' or """.

````python
"""This is also a
dd
adad
asdasdas
asdsad
perfect example of
multi-line comments"""
````

# DocString in python

Docstring is short for documentation string.

It is a string that occurs as the first statement in a module, function, class, or method definition.

````python
def double(num):
    """
    function to double the number
    """
    return 2 * num

print(double(10))
````

````python
print(double.__doc__) #Docstring is available to us as the attribute __doc__ of the function
````

# Python Indentation

2. A code block (body of a function, loop etc.) starts with indentation and ends with the first unindented line. The amount of indentation is up to you, but it must be consistent throughout that block.

## Comments, Docstrings & Indentation — Quick Reference

| Feature | Syntax | Purpose |
|---------|--------|---------|
| **Single-line comment** | `# note` | Short explanations, ignored by Python |
| **Multi-line comment** | several `#` lines | Block of notes |
| **String literal (sometimes used as comment)** | `""" ... """` | Not a real comment — Python evaluates it but discards it |
| **Docstring** | `""" ... """` as **first** statement in a def/class/module | Becomes `__doc__`, used by `help()` |
| **Indentation** | 4 spaces (PEP 8) | Defines code blocks (replaces `{}` of other languages) |
| **Line continuation** | trailing `\` or `( ... )` | Split a long statement across lines |
| **Multiple statements per line** | `a = 1; b = 2` | Allowed but discouraged |

> **Rule of thumb:** be consistent — don't mix tabs and spaces, and don't change indent width inside the same block.

---
## Good vs. Bad Comments

```python
# BAD — restates the code
x = x + 1   # add 1 to x

# GOOD — explains the WHY
x = x + 1   # compensate for zero-based index

# BAD — outdated or misleading comment
# returns the user's age (actually returns birth year)
def get_user(): ...

# GOOD — docstring describing intent, args, return
def get_user(user_id):
    """Look up a user by id and return the User object."""
```
