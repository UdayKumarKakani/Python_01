# Exception Handling:

---

When an error occurs, or exception as we call it, Python will normally stop and generate an error message.

These exceptions can be handled using the try statement:

````
# The try block will generate an exception, because x is not defined:

try:
  print(x)
except:
  print("An exception occurred")
````

Since the try block raises an error, the except block will be executed.

## Many Exceptions

---


You can define as many exception blocks as you want, e.g. if you want to execute a special block of code for a special kind of error:

````
# Print one message if the try block raises a NameError and another for other errors:

try:
  print(x)
except NameError:
  print("Variable x is not defined")
except:
  print("Something else went wrong")
````

## Else
You can use the else keyword to define a block of code to be executed if no errors were raised:

Example
In this example, the try block does not generate any error:

````
try:
  print("Hello");
except:
  print("Something went wrong")
else:
  print("Nothing went wrong")
````

## Finally
The finally block, if specified, will be executed regardless if the try block raises an error or not.

Example


````
try:
  print(x)
except:
  print("Something went wrong")
finally:
  print("The 'try except' is finished")
````

````
#This can be useful to close objects and clean up resources:

# Try to open and write to a file that is not writable:

#The try block will raise an error when trying to write to a read-only file:

try:
  file = open("demo.txt","r")
  file.write("Lorum Ipsum")
except:
  print("Something went wrong when writing to the file")

finally:
  file.close()

#The program can continue, without leaving the file object open
````

````
file = open("demo.txt","w")
file.write("Lorum Ipsum")

file.close()
````

## Common Exception Types

| Exception | When it occurs | Example |
|-----------|---------------|--------|
| `ValueError` | Wrong value for the type | `int("hello")` |
| `TypeError` | Wrong data type in operation | `"a" + 1` |
| `ZeroDivisionError` | Division by zero | `10 / 0` |
| `IndexError` | List index out of range | `lst[99]` on a 3-item list |
| `KeyError` | Dictionary key not found | `d["missing"]` |
| `FileNotFoundError` | File doesn't exist | `open("no.txt")` |
| `AttributeError` | Object has no such attribute | `"hello".push("x")` |
| `NameError` | Variable used before being defined | `print(x)` when x doesn't exist |

## Catching the Exception Object with `as e`

Adding `as e` after the exception type gives you access to the error message:

```python
try:
    result = 10 / 0
except ZeroDivisionError as e:
    print(f"Error caught: {e}")
# Output: Error caught: division by zero
```

````
try:
    result = 10 / 0
except ZeroDivisionError as e:
    print(f"ZeroDivisionError: {e}")

try:
    x = int("hello")
except ValueError as e:
    print(f"ValueError: {e}")

try:
    my_list = [1, 2, 3]
    print(my_list[10])
except IndexError as e:
    print(f"IndexError: {e}")
````

## `raise` — Manually Raising Exceptions

You can raise your own exceptions to enforce rules in your code:

```python
def check_age(age):
    if age < 0:
        raise ValueError("Age cannot be negative!")
    return age

try:
    check_age(-5)
except ValueError as e:
    print(f"Error: {e}")
```

````
def check_age(age):
    if age < 0:
        raise ValueError("Age cannot be negative!")
    if age > 150:
        raise ValueError("Age seems unrealistic!")
    return age

for test_age in [25, -5, 200]:
    try:
        print(f"Age {test_age} is valid: {check_age(test_age)}")
    except ValueError as e:
        print(f"Error for age {test_age}: {e}")
````
