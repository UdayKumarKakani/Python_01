# Virtual Environments & Dependency Management

---

A **virtual environment** is an isolated Python installation for a single project. It has its own `python` binary and its own set of installed packages, completely separate from your system Python.

> **Note:** Most commands in this notebook are **shell commands** meant to be run in your **terminal**, not inside Jupyter. They're shown as markdown code blocks for reference.

## Why Virtual Environments Matter

- **Isolation** — each project has its own dependencies, untouched by other projects
- **No version conflicts** — Project A can use `requests 2.20` while Project B uses `requests 2.31`
- **Reproducibility** — anyone can recreate the exact same environment on another machine
- **Clean system Python** — you never pollute your global Python install with project-specific packages

## The Dependency Hell Problem

Imagine you have two projects on the same laptop:

| Project | Needs |
|---------|-------|
| **Project A** (old web scraper) | `requests==2.20` |
| **Project B** (new API client) | `requests==2.31` |

If you install both globally, the second `pip install` **overwrites** the first. One project breaks.

Virtual environments solve this by giving each project its own private `site-packages` folder.

---
## `venv` — The Built-in Tool

Python ships with `venv` out of the box. No installation needed.

### Create a virtual environment
```bash
python -m venv .venv
```
This creates a folder called `.venv/` in your project directory.

### Activate it

**Mac / Linux:**
```bash
source .venv/bin/activate
```

**Windows (PowerShell / cmd):**
```bash
.venv\Scripts\activate
```

Once active, your prompt usually changes to show `(.venv)` at the front.

### Deactivate
```bash
deactivate
```

### What's inside `.venv/`?

- `bin/` (or `Scripts/` on Windows) — the isolated `python` and `pip` executables
- `lib/python3.x/site-packages/` — where your installed packages live
- `pyvenv.cfg` — config file pointing to the base Python

> **Never commit `.venv/` to git.** It's machine-specific and can be huge. Add it to `.gitignore` instead.

````python
# You can check which python is active from inside Jupyter:
import sys
print(sys.executable)
````

---
## `pip` — The Package Installer

`pip` installs packages from [PyPI](https://pypi.org) into the **currently active** environment.

| Command | What it does |
|---------|-------------|
| `pip install requests` | Install latest version |
| `pip install requests==2.31.0` | Install a pinned version |
| `pip install "requests>=2.20,<3.0"` | Install within a version range |
| `pip install -r requirements.txt` | Install everything in a requirements file |
| `pip freeze > requirements.txt` | Save current packages + versions to a file |
| `pip list` | Show installed packages |
| `pip show requests` | Show details about a package |
| `pip uninstall requests` | Remove a package |

````python
# You can run shell commands from Jupyter using the ! prefix:
!pip list | head -10
````

---
## `requirements.txt` Format

A simple text file listing dependencies — one per line. Example:

```text
requests==2.31.0
numpy>=1.24.0
pandas
flask==3.0.0
python-dotenv>=1.0.0
```

**Operators:**
- `==` — exact pin (recommended for production)
- `>=` — minimum version
- `<=` — maximum version
- no operator — any version (latest)

Generate one from your current environment:
```bash
pip freeze > requirements.txt
```

---
## `uv` — The Modern, Fast Alternative

[`uv`](https://github.com/astral-sh/uv) is a Rust-based Python package manager that's **10–100x faster** than pip. It's a drop-in replacement for `pip` and `venv`.

### Install uv
```bash
# Mac / Linux
curl -LsSf https://astral.sh/uv/install.sh | sh

# or via pip
pip install uv
```

### Common uv commands
```bash
uv venv                          # create .venv (faster than python -m venv)
uv pip install requests          # install a package
uv pip install -r requirements.txt
uv pip freeze > requirements.txt
uv pip sync requirements.txt     # make env match the file exactly
```

**When to use uv over pip:**
- Large projects with many dependencies (huge speed gains)
- CI/CD pipelines (faster builds)
- You want strict reproducibility (`uv pip sync`)

---
## `conda` / Anaconda — Brief Mention

[Conda](https://docs.conda.io/) is a different kind of package manager — it handles **non-Python dependencies** (C/C++ libraries, CUDA, BLAS, etc.) too. Popular in data science.

### When conda shines
Heavy native dependencies: **NumPy, SciPy, PyTorch, TensorFlow, OpenCV** — conda often has pre-built binaries that just work.

### Basic commands
```bash
conda create -n myenv python=3.11
conda activate myenv
conda install numpy pandas
conda deactivate
```

> **Warning:** Don't mix `pip install` and `conda install` haphazardly in the same env — they can disagree about what's installed and break things. If you must mix, install conda packages **first**, then pip on top.

---
## Reference Table — venv vs uv vs conda

| Tool | Speed | Best for | Create command |
|------|-------|----------|---------------|
| **venv** | Standard | Any Python project — always available | `python -m venv .venv` |
| **uv** | 10–100x faster | Modern projects, CI/CD, large dep trees | `uv venv` |
| **conda** | Slower | Data science with heavy native libs (PyTorch, SciPy) | `conda create -n myenv python=3.11` |

**Rule of thumb:** start with `venv` (or `uv` for speed). Reach for `conda` only when native deps cause you pain.

---
## Best Practices

- **One venv per project** — never share environments across projects
- **Always pin versions in production** — use `==` so deploys are reproducible
- **Add to `.gitignore`:**
  ```text
  .venv/
  __pycache__/
  *.pyc
  .env
  ```
- **Use `python -m pip install ...`** instead of bare `pip install ...` — this guarantees you're using pip from the *currently active* Python and avoids PATH surprises
- **Commit `requirements.txt`**, not `.venv/`
- **Activate the venv before working** — every new terminal needs activation
