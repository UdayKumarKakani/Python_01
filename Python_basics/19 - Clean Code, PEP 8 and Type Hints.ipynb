# Clean Code, PEP 8 & Type Hints

---

> "Code is read more than it's written."

In production and on teams, clean code matters more than clever code. It saves review time, reduces bugs, and makes onboarding painless. Python gives us an official style guide (**PEP 8**) and optional **type hints** to make intent crystal clear.

## PEP 8 — Python's Official Style Guide

**PEP 8** (Python Enhancement Proposal 8) is the style guide that the Python community follows. It covers naming, spacing, layout, imports, and more.

You don't need to memorize it — tools like `ruff` and `black` enforce it automatically. But knowing the basics helps you write code that *looks* like Python.

---
## Naming Conventions

| Item | Convention | Example |
|------|-----------|--------|
| Variable | `snake_case` | `user_name` |
| Function | `snake_case` | `get_user()` |
| Constant | `UPPER_SNAKE_CASE` | `MAX_RETRIES = 3` |
| Class | `PascalCase` | `UserAccount` |
| Private | leading underscore | `_internal_helper` |
| Module / file | lowercase | `my_module.py` |

> **Rule of thumb:** if it's a value, use `snake_case`. If it's a type/class, use `PascalCase`. If it never changes, use `UPPER_SNAKE_CASE`.

````python
# Naming conventions in practice
MAX_RETRIES = 3              # constant
user_name = "Alice"          # variable

def get_user(user_id):       # function
    return {"id": user_id, "name": user_name}

class UserAccount:           # class
    def __init__(self, name):
        self.name = name
        self._token = None   # private (by convention)

print(get_user(1))
print(UserAccount("Bob").name)
````

---
## Indentation, Spacing & Line Length

- **4 spaces** per indent level (never tabs)
- **79 characters** max line length (or **88** if you use `black` / `ruff`)
- **2 blank lines** between top-level functions and classes
- **1 blank line** between methods inside a class
- **Spaces around operators:** `x = 1 + 2`, not `x=1+2`
- **No space inside parens:** `func(x, y)`, not `func( x, y )`
- **Space after comma:** `[1, 2, 3]`, not `[1,2,3]`

---
## Import Order (PEP 8)

Imports go at the **top** of the file, grouped in this order with a blank line between groups:

1. **Standard library** (e.g. `os`, `sys`, `json`)
2. **Third-party packages** (e.g. `numpy`, `requests`)
3. **Local / project modules**

```python
import os
import sys
from pathlib import Path

import numpy as np
import requests

from my_project.utils import helper
```

> Tools like `ruff` / `isort` sort imports automatically.

## Bad vs Good — Side by Side

Same logic, very different readability.

````python
# BAD: cramped, inconsistent, unclear names
def Calc(x,y):
  z=x+y
  if z>10:return "big"
  else:return "small"

print(Calc(3,4))
print(Calc(8,5))
````

````python
# GOOD: PEP 8 compliant, clear names, proper spacing
def classify_sum(a, b):
    total = a + b
    if total > 10:
        return "big"
    return "small"

print(classify_sum(3, 4))
print(classify_sum(8, 5))
````

## Docstrings

A **docstring** is a string literal placed right after a `def`, `class`, or at the top of a module. It documents *what* the code does.

Popular formats: **Google style** (clean, readable), NumPy style, reStructuredText.

Access docstrings via `help(func)` or `func.__doc__`.

````python
def add(a: int, b: int) -> int:
    """Add two integers.

    Args:
        a: First number.
        b: Second number.

    Returns:
        Sum of a and b.
    """
    return a + b

help(add)
print("---")
print(add.__doc__)
````

## Type Hints (PEP 484)

Type hints declare *expected* types of variables, function parameters, and return values.

```python
def greet(name: str) -> str:
    return f"Hello, {name}"

count: int = 0
```

**Collections** (Python 3.9+):
- `list[int]` — list of ints
- `dict[str, int]` — dict mapping strings to ints
- `tuple[int, str]` — tuple of (int, str)

**Optional / Union**:
- `Optional[int]` — int **or** None (`from typing import Optional`)
- `int | None` — same thing, modern syntax (Python 3.10+)
- `int | str` — union of types
- `Any` — escape hatch, accept anything (`from typing import Any`)

> **Hints are HINTS.** Python does **not** enforce them at runtime. Use **mypy** or **pyright** for static checking.

````python
from typing import Optional, Any

def greet(name: str) -> str:
    return f"Hello, {name}"

def total(nums: list[int]) -> int:
    return sum(nums)

def get_age(user: dict[str, Any]) -> Optional[int]:
    return user.get("age")

print(greet("Alice"))
print(total([1, 2, 3, 4]))
print(get_age({"name": "Bob", "age": 30}))
print(get_age({"name": "Eve"}))
````

## The `if __name__ == "__main__":` Idiom

When you run a Python file directly, Python sets the special variable `__name__` to `"__main__"`. When the file is **imported** by another module, `__name__` becomes the module's name instead.

This idiom lets a file behave **both** as a reusable library *and* as a runnable script:

```python
def greet(name):
    return f"Hello, {name}"

def main():
    print(greet("World"))

if __name__ == "__main__":
    main()
```

Now `import this_file` won't trigger `main()` — only running `python this_file.py` will.

````python
def greet(name: str) -> str:
    return f"Hello, {name}"

def main() -> None:
    print(greet("World"))
    print(f"__name__ is: {__name__}")

if __name__ == "__main__":
    main()
````

## Modern Tooling

You don't have to format code by hand. These tools do it for you:

- **`ruff`** — extremely fast linter + formatter. Replaces `flake8`, `isort`, and most of `black` in one tool.
- **`black`** — opinionated auto-formatter. Zero config.
- **`mypy`** / **`pyright`** — static type checkers that validate your type hints.

Common commands:

```bash
ruff check .         # lint the project
ruff format .        # auto-format the project
black myfile.py      # format one file with black
mypy myfile.py       # check types
```

---
## Tooling Reference

| Tool | Type | What it does |
|------|------|--------------|
| `ruff` | Linter + formatter | Catches style issues, sorts imports, formats code (very fast) |
| `black` | Formatter | Reformats code to a strict, consistent style |
| `isort` | Import sorter | Groups and sorts imports per PEP 8 |
| `flake8` | Linter | Older PEP 8 + bug checker (mostly replaced by ruff) |
| `mypy` | Type checker | Validates type hints against actual usage |
| `pyright` | Type checker | Microsoft's fast type checker (used by Pylance / VS Code) |
