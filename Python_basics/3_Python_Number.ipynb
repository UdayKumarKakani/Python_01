#Python Numbers

**There are three numeric types in Python:**

*   Int
*   Float
*   Complex

Variables of numeric types are created when you assign a value to them:

---





````
x = 1    # int
y = 2.8  # float
z = 1j   # complex
print(type(x))
print(type(y))
print(type(z))
````

**Int**

Int, or integer, is a whole number, positive or negative, without decimals, of unlimited length.

---



````
x = 1
y = 35656222554887711
z = -3255522

print(type(x))
print(type(y))
print(type(z))
````

**Float**

Float, or "floating point number" is a number, positive or negative, containing one or more decimals.

---



````
x = 1.1
y = 1.0
z = -35.59

print(type(x))
print(type(y))
print(type(z))
````

Float can also be scientific numbers with an "e" to indicate the power of 10.

````
x = 35e3
y = 12E4
z = -87.7e100

print(type(x))
print(type(y))
print(type(z))
````

**Complex**

Complex numbers are written with a "j" as the imaginary part:

---



````
x = 3+5j
y = 5j
z = 1-5j

print(x +y +z)
print(type(y))
print(type(z))
````

## Type Conversion

Python lets you convert between numeric types using built-in functions.
This is essential when handling user input (which is always a **string** by default).

| Function | Converts to | Example |
|----------|-------------|--------|
| `int(x)` | Integer | `int("42")` → `42`, `int(3.9)` → `3` (truncates) |
| `float(x)` | Float | `float("3.14")` → `3.14`, `float(5)` → `5.0` |
| `str(x)` | String | `str(100)` → `"100"` |
| `complex(x)` | Complex | `complex(3)` → `(3+0j)` |

> **Important:** `int()` always **truncates** (removes decimals) — it does NOT round.
> `int(3.9)` → `3`, not `4`. Use `round()` to round.

````
# Type conversion examples
print(int("42"))        # string → int: 42
print(float("3.14"))    # string → float: 3.14
print(str(100))         # int → string: '100'
print(int(3.9))         # float → int (truncates): 3
print(round(3.9))       # rounds: 4
print(type(str(100)))   # confirms it's now a string
````

## Useful Built-in Math Functions

Python has several built-in functions for common math operations — no import needed:

| Function | Description | Example |
|----------|-------------|--------|
| `abs(x)` | Absolute value | `abs(-7)` → `7` |
| `round(x, n)` | Round to n decimal places | `round(3.567, 2)` → `3.57` |
| `max(a, b, ...)` | Largest value | `max(3, 7, 2)` → `7` |
| `min(a, b, ...)` | Smallest value | `min(3, 7, 2)` → `2` |
| `pow(base, exp)` | Power | `pow(2, 10)` → `1024` |
| `sum(iterable)` | Sum of a list | `sum([1,2,3,4])` → `10` |

For advanced math (sqrt, sin, cos, etc.) use the `math` module: `import math`

````
print(abs(-7))          # 7
print(round(3.567, 2))  # 3.57
print(max(3, 7, 2))     # 7
print(min(3, 7, 2))     # 2
print(pow(2, 10))       # 1024
print(sum([1,2,3,4,5])) # 15

import math
print(math.sqrt(144))   # 12.0
print(math.pi)          # 3.14159...
````
