**String Formatting**

Python uses C-style string formatting to create new, formatted strings. The "%" operator is used to format a set of variables enclosed in a "tuple" (a fixed size list), together with a format string, which contains normal text together with "argument specifiers", special symbols like "%s" and "%d".

Let's say you have a variable called "name" with your user name in it, and you would then like to print(out a greeting to that user.)

---

**Example**:  This prints out "Hello, John!"

````python
name = "John"
title='Mr.'
print("Hello, %s %s !" % (title,name))
````

**Example ** : "This prints out "John is 23 years old."

---



````python
name = "John"
age = 23
print("%s is %d years old is." % (name, age))
````

Any object which is not a string can be formatted using the %s operator as well. The string which returns from the "repr" method of that object is formatted as the string.

---


For example:

````python
mylist = [1,2,3]
print("A list: %s" % mylist)
````

**Here are some basic argument specifiers you should know:**

---



**%s** - String (or any object with a string representation, like numbers)

**%d** - Integers

**%f** - Floating point numbers upto 6 decimal points

**%.<number of digits>f** - Floating point numbers with a fixed amount of digits to the right of the dot.

**%x/%X -** Integers in hex representation (lowercase/uppercase)

## f-Strings (Modern String Formatting — Python 3.6+)

f-strings are the **preferred modern way** to embed variables inside strings. Just prefix the string with `f` and put variable names inside `{}`.

```python
name = "Priya"
age = 22
print(f"My name is {name} and I am {age} years old.")
# Output: My name is Priya and I am 22 years old.
```

You can also do math inside `{}`:
```python
price = 150
qty = 3
print(f"Total: ₹{price * qty}")
# Output: Total: ₹450
```

> **Tip:** f-strings are cleaner than `%` formatting or `.format()`. Use them whenever possible.

````python
name = "Priya"
age = 22
city = "Bangalore"
print(f"My name is {name}, I am {age} years old, from {city}.")

price = 150
qty = 3
print(f"Total: ₹{price * qty}")

pi = 3.14159
print(f"Pi rounded to 2 decimals: {pi:.2f}")  # :.2f formats to 2 decimal places
````

## Escape Characters

Escape characters let you include special characters inside strings:

| Escape | Meaning | Example |
|--------|---------|--------|
| `\n` | New line | `"Hello\nWorld"` prints on 2 lines |
| `\t` | Tab (indent) | `"Name:\tRaj"` |
| `\\` | Literal backslash | `"C:\\Users"` |
| `\"` | Double quote inside double-quoted string | `"He said \"Hello\""` |
| `\'` | Single quote inside single-quoted string | `'It\'s Python'` |

````python
print("Hello\nWorld")          # newline
print("Name:\tRaj")            # tab
print("C:\\Users\\Documents")  # literal backslash
print("He said \"Hello!\"")     # quotes inside quotes
print('It\'s Python')          # apostrophe in single-quoted string
````
