# Python Keywords

Keywords are the reserved words in python

We can't use a keyword as variable name, function name or any other identifier

Keywords are case sentive

````python
#Get all keywords in python 3.6

import keyword

print(keyword.kwlist)

print("Total number of keywords ", len(keyword.kwlist))
````

# Identifiers

Identifier is the name given to entities like class, functions, variables etc. in Python. It helps differentiating one entity from another.

Rules for Writing Identifiers:

1. Identifiers can be a combination of letters in lowercase (a to z) or uppercase (A to Z) or digits (0 to 9) or an underscore (_).

2. An identifier cannot start with a digit. 1variable is invalid, but variable1 is perfectly fine.

3. Keywords cannot be used as identifiers.

````python
a1= 1
print(_abcABC019)
````

 We cannot use special symbols like !, @, #, $, % etc. in our identifier.

````python
a@ = 10             #can't use special symbols as an identifier
````

---
## Python Keywords Reference

| Category | Keywords |
|----------|----------|
| **Logical values** | `True`, `False`, `None` |
| **Logical operators** | `and`, `or`, `not` |
| **Conditionals** | `if`, `elif`, `else` |
| **Loops** | `for`, `while`, `break`, `continue`, `pass` |
| **Functions / classes** | `def`, `return`, `lambda`, `class`, `yield` |
| **Exceptions** | `try`, `except`, `finally`, `raise`, `assert` |
| **Imports / scope** | `import`, `from`, `as`, `global`, `nonlocal` |
| **Async / context** | `async`, `await`, `with` |
| **Identity / membership** | `is`, `in`, `del` |

> Use `help("keywords")` in the interpreter to see them all, or `keyword.iskeyword("if")` to test a word.

---
## Identifier Naming Conventions (PEP 8)

| Style | Example | Used for |
|-------|---------|----------|
| `snake_case` | `student_name` | Variables, functions |
| `PascalCase` | `StudentRecord` | Classes |
| `UPPER_CASE` | `MAX_LIMIT` | Constants |
| `_leading_underscore` | `_internal` | Private/internal use |
| `__double_underscore` | `__private` | Strongly private (name-mangled) |
| `__dunder__` | `__init__` | Reserved for Python (don't invent your own) |

> Identifiers are **case-sensitive**: `Age`, `age`, and `AGE` are three different names.
