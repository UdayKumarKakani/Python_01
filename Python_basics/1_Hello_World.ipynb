## What is `print()`?

`print()` is Python's built-in function to display output on the screen. Whatever you put inside the parentheses gets shown to the user.

### Key Concepts
- **String**: Text enclosed in single `'...'` or double `"..."` quotes — e.g., `"Hello World"`
- **Comment**: A line starting with `#` is ignored by Python — use it to explain your code
- **Case-sensitivity**: `Print` and `print` are different in Python — always use lowercase
- **Indentation**: Python uses spaces/tabs to define blocks of code (you'll see this in loops and functions)

```python
# This is a comment — Python ignores this line
print("Hello!")      # prints: Hello!
print('Python')      # single quotes work too
print(42)            # numbers don't need quotes
print(3 + 5)         # prints the result: 8
```

````python
# This is a comment
print("Hello World")       # basic string
print('Single quotes too') # single quotes work
print(42)                  # printing a number
print(3 + 5)               # printing a math result
print("Hello" + " " + "World")  # joining strings with +
````
