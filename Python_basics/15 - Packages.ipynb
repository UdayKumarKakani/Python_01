# Package

Packages are a way of structuring Python’s module namespace by using “dotted module names”.

A directory must contain a file named __init__.py in order for Python to consider it as a package. This file can be left empty but we generally place the initialization code for that package in this file.

## Package vs Module — Quick Reference

| Concept | What it is | Example |
|---------|-----------|---------|
| **Module** | A single `.py` file | `math_operations.py` |
| **Package** | A directory of modules with `__init__.py` | `my_utilities/` |
| **Sub-package** | A package inside a package | `my_utilities/strings/` |
| **`__init__.py`** | Marker file — tells Python "this is a package" | Often empty, can re-export names |

### Common Import Patterns

| Syntax | Meaning |
|--------|---------|
| `import my_utilities` | Import whole package (runs `__init__.py`) |
| `from my_utilities import math_operations` | Import a specific module |
| `from my_utilities.math_operations import add` | Import a single function |
| `from my_utilities import *` | Import everything listed in `__all__` |

---
## The Role of `__init__.py`

The `__init__.py` file makes a directory a **package**. It serves three purposes:

1. **Marker** — Python recognizes the folder as a package
2. **Initialization** — code in it runs when the package is first imported
3. **Re-exporting** — lets you "flatten" the API:

```python
# my_utilities/__init__.py
from .math_operations import add, subtract
from .string_operations import uppercase

# Now users can do:
#   from my_utilities import add
# instead of:
#   from my_utilities.math_operations import add
```

> **Modern note:** Python 3.3+ supports **namespace packages** (no `__init__.py` needed) but for regular packages it's still best practice to include one.
