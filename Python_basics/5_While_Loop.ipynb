#Python While Loops

**The while Loop**

With the while loop we can execute a set of statements **as long as a condition is true.**

The while loop requires relevant variables to be ready, in this example we need to define an indexing variable, i, which we set to 1.


---

Note: remember to increment i, or else the loop will continue forever.

````
i = 1
while i < 6:
  print(i)
  i += 1
````

**The break Statement**

With the break statement we can stop the loop even if the while condition is true:

---



````
i = 1

while i < 6:
  print(i)

  if i ==3:
    break
  i += 1
````

**The continue Statement**

With the continue statement we can stop the current iteration, and continue with the next:

---



````
i = 0
while i < 6:
  i += 1
  if i == 3:
    continue
  print(i)
````

---
## ⚠️ Infinite Loop Warning

A while loop whose condition **never becomes False** runs forever — this will crash or freeze your program:

```python
# DANGER — infinite loop!
i = 1
while i < 10:
    print(i)
    # Forgot: i += 1  ← loop never ends!
```

**Always check:** Is the loop variable being updated inside the loop?

---
## while-else

Just like `for`, a `while` loop can have an `else` block.  
The `else` runs **only if the loop completed normally** (i.e., was not stopped by `break`):


````
count = 0
while count < 5:
    print(count)
    count += 1
else:
    print("Loop finished normally!")

print()
# With break — else does NOT run:
count = 0
while count < 5:
    if count == 3:
        break
    print(count)
    count += 1
else:
    print("This won't print because we used break")
````

---
## When to Use `while` vs `for`

| | `for` loop | `while` loop |
|---|---|---|
| Use when | You know the number of iterations | You loop until a condition changes |
| Examples | Iterate over a list, `range(n)` | User input validation, game loops, retry logic |
| Risk | Generally safe | Risk of infinite loop if condition never changes |
