# JSON and CSV File Handling

---

Real-world programs rarely work with isolated values — they read and write **structured data**.  
The two most common formats for this are **JSON** and **CSV**:

- **JSON** (JavaScript Object Notation) — nested key/value data, used by almost every web API
- **CSV** (Comma-Separated Values) — flat tabular data, used by Excel, databases, and data pipelines

Python has built-in modules for both — `json` and `csv` — no installation needed.

## Why JSON and CSV Matter

- **Universal** — every language and tool understands them
- **Human-readable** — open them in any text editor
- **Lightweight** — no databases or servers required
- **Perfect for config, logs, exports, and API responses**

---
## Working with JSON

Import the built-in `json` module:

````python
import json
````

### `json.dumps()` — Python dict → JSON string

`dumps` = **dump string**. Converts a Python object into a JSON-formatted string.

````python
person = {"name": "Alice", "age": 25, "is_student": True, "hobbies": ["reading", "coding"]}

json_string = json.dumps(person)
print(json_string)
print(type(json_string))
````

### `json.loads()` — JSON string → Python dict

`loads` = **load string**. The reverse of `dumps`.

````python
data = json.loads(json_string)
print(data)
print(type(data))
print(data["name"])
````

### `json.dump()` — Write Python object to a file

Notice the missing `s` — `dump` writes directly to a file object.

````python
person = {"name": "Alice", "age": 25, "hobbies": ["reading", "coding"]}

with open("person.json", "w") as f:
    json.dump(person, f)

print("Saved person.json")
````

### `json.load()` — Read JSON from a file

````python
with open("person.json", "r") as f:
    loaded = json.load(f)

print(loaded)
print(loaded["hobbies"])
````

### Pretty-printing with `indent` and `sort_keys`

By default JSON is written on a single line. Pass `indent=2` for readable output, and `sort_keys=True` to alphabetize keys.

````python
person = {"name": "Alice", "age": 25, "hobbies": ["reading", "coding"]}

pretty = json.dumps(person, indent=2, sort_keys=True)
print(pretty)
````

---
## Python ↔ JSON Type Mapping

| Python              | JSON      |
|---------------------|-----------|
| `dict`              | object    |
| `list`, `tuple`     | array     |
| `str`               | string    |
| `int`, `float`      | number    |
| `True` / `False`    | true / false |
| `None`              | null      |

> **Note:** JSON only supports these types. You **cannot** dump a `set`, a `datetime`, or a custom class directly — you'll need to convert them first.

### Worked Example — Save & Load a List of Students

````python
students = [
    {"name": "Alice", "age": 20, "marks": 88},
    {"name": "Bob",   "age": 22, "marks": 75},
    {"name": "Charlie", "age": 21, "marks": 92},
]

# Save to file
with open("students.json", "w") as f:
    json.dump(students, f, indent=2)

# Load back
with open("students.json", "r") as f:
    loaded_students = json.load(f)

for s in loaded_students:
    print(s["name"], "->", s["marks"])
````

## Working with CSV

CSV files store **tabular data** — rows and columns separated by commas.  
Import the built-in `csv` module:

````python
import csv
````

### Writing with `csv.writer` — rows as lists

> **Tip:** Always open CSV files with `newline=""` to avoid extra blank lines on Windows.

````python
rows = [
    ["name", "age", "grade"],   # header
    ["Alice", 20, "A"],
    ["Bob", 22, "B"],
    ["Charlie", 21, "A"],
]

with open("grades.csv", "w", newline="") as f:
    writer = csv.writer(f)
    writer.writerows(rows)

print("Saved grades.csv")
````

### Reading with `csv.reader` — each row is a list

````python
with open("grades.csv", "r", newline="") as f:
    reader = csv.reader(f)
    for row in reader:
        print(row)
````

### `csv.DictWriter` / `csv.DictReader` — rows as dicts

Using dicts is cleaner when your CSV has headers — you can refer to columns by name instead of index.

````python
students = [
    {"name": "Alice", "age": 20, "grade": "A"},
    {"name": "Bob", "age": 22, "grade": "B"},
    {"name": "Charlie", "age": 21, "grade": "A"},
]

# Write
with open("students.csv", "w", newline="") as f:
    writer = csv.DictWriter(f, fieldnames=["name", "age", "grade"])
    writer.writeheader()
    writer.writerows(students)

# Read
with open("students.csv", "r", newline="") as f:
    reader = csv.DictReader(f)
    for row in reader:
        print(row["name"], "->", row["grade"])
````

---
## When to use JSON vs CSV

| Use **JSON** when... | Use **CSV** when... |
|---|---|
| Data is **nested** (lists inside dicts) | Data is **flat / tabular** |
| Working with **web APIs** | Working with **spreadsheets** |
| Values have **mixed types** | All rows have the **same columns** |
| Order of keys doesn't matter | Order of columns matters |
| You need **structure** | You need **simplicity** |
