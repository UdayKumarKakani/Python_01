# Day 2 — Assignments


## Exercise 1 — Record yourself on 5 questions

Pick 5 questions from today. Record 2-minute answers to each on your phone. Watch back. Score yourself out of 10: clarity, numbers used, tradeoffs mentioned, closed with monitoring.


## Exercise 2 — Pin 6 repos with polished READMEs

Pin 6 repos on GitHub. For each, write a README using the class template. Add a screenshot or GIF.


## Exercise 3 — 1-page resume

Draft an AI-eng resume from `resume_template.md`. Every project bullet must have a number. Have a friend proofread.


## Exercise 4 — Run the class mock yourself

45-min timer. Whiteboard the sales-team assistant design end-to-end. Record. Watch back.


## Stretch — 5 mocks in a week

Run 5 different mocks (see class list) over 5 days. Track a self-score. **Pair mocks with a friend** if possible — highest-leverage prep activity.
