## Part 1 — 20 real interview questions

**Bucket A — Concepts** (10)

**Q1. RAG vs fine-tuning?** RAG adds knowledge at query time (retrieve → augment → generate) — great for changing / private facts. Fine-tuning changes weights — great for style / format / narrow skills. Exhaust prompting + RAG before fine-tuning.

**Q2. What is an embedding?** Dense vector where semantically similar text points similar directions. Enables semantic search, dedup, clustering. 384–3072 dims.

**Q3. Cosine vs Euclidean vs dot product?** Cosine is default for text (direction). Dot on normalized vectors = cosine. Euclidean rarely used for text.

**Q4. Chunking — why does size matter?** 200–800 tokens typical. Too big → vector vague. Too small → fragmented context. Recursive splitter (paragraphs → sentences) standard.

**Q5. HNSW?** Graph-based ANN. Multi-layer graph, greedy traversal. Small recall drop, huge speedup vs brute force.

**Q6. Cross-encoder reranker?** Takes `(query, doc)` together, scores relevance. More accurate than bi-encoder, too slow to run over all docs. Retrieve 20-50 with bi-encoder, rerank to top-5.

**Q7. LoRA?** Freeze base weights, train tiny low-rank patches (A×B, rank 8-64). ~1% params, similar quality on style tasks. Adapters ~50 MB, swappable.

**Q8. Why LLMs hallucinate + mitigations?** They optimize likely-sounding text, not truth. Mitigations: RAG grounding, "if not in context say I don't know", refuse on weak retrieval, reranking, user feedback.

**Q9. Function calling?** LLM outputs structured JSON matching your schemas → your code runs the tool → sends back result → loop. Backbone of agents.

**Q10. Evaluating a RAG?** Start with 20 manual Q&A pairs, score correct/partial/wrong. Automate with LLM-as-judge for faithfulness. Track in Langfuse. Ship only if better than baseline.


**Bucket B — Systems** (7)

**Q11. Chatbot over 100k internal docs?** Ingestion nightly. pgvector or Chroma at that scale. Hybrid retrieve → rerank → LLM. SSO auth. Langfuse tracing. Per-user daily cap. p95 SLO ~4s. Cost ~$X/month.

**Q12. Quality dropped — how debug?** Langfuse: which step degraded? Diff recent deploys. Manual eval on 20 known-good queries. Check ingestion for a bad batch. Roll back if the delta is clear.

**Q13. Prompt injection defenses?** Wrap user input in explicit tags. Regex-filter obvious attacks. Sanitize retrieved content. Never let LLM output run destructive actions without approval. Log suspicious inputs.

**Q14. Bill is $10k/mo — cut it?** Semantic cache (biggest). Prompt caching. Route classification to smaller model. `max_tokens` cap. Shorter system prompts. Self-host only after all the above.

**Q15. Streaming end to end?** LLM API returns SSE tokens → FastAPI `StreamingResponse` → browser `EventSource` or `fetch` stream. First-token latency is what users perceive.

**Q16. When OpenAI over open-source?** OpenAI: no infra, best quality out of box, small volume. Llama: cost-sensitive, data can't leave VPC, want to fine-tune. Often both, A/B tested.

**Q17. Design an autonomous business agent?** Perception (trigger) → planner (LLM) → tools (email, DB, HTTP) → executor with retries + budget → HITL for destructive actions → memory → observability. Cost cap per run. Log every step.


**Bucket C — Behavioral** (3)

**Q18. Tell me about a project you're proud of.** STAR: Situation → Task → Action → Result. Pick a Section 6/7/8 capstone. **Numbers matter**: "reduced hallucination rate from X% to Y% after adding reranking + distance threshold."

**Q19. Wrong technical decision?** Interviewers want humility + growth. Pick a real one. "I built X thinking it would scale; didn't past Y load. Rewrote using Z. Now I default to profiling before optimizing."

**Q20. Why AI engineering (vs backend / DS / research)?** Show you know the difference:
- **AI eng** = ship LLM-powered features end-to-end
- **DS** = analyze data, models for insights
- **ML research** = train new models, publish

Then why *you* picked AI eng — usually: "I love building products, and modern AI lets one engineer ship what used to need a research team."


### The meta-answer

Regardless of question, lean on three things and you'll sound senior:

- **Number**: "About $X per request." "p95 ~ Y ms."
- **Tradeoff**: "I picked X because A. If B mattered more, I'd pick Y."
- **Monitoring**: "Trace in Langfuse; alert if p95 or cost drift."

Say those three in every answer.

## Part 3 — Full mock system-design walkthrough

**Prompt (real 2026 interview):**

> *"Design an AI assistant that helps our sales team quickly answer questions about our product using our internal Confluence, PDFs, and past support tickets. ~500,000 docs. 300 sales people. Search + Q&A."*

We walk this in the 5 steps.


### Step 1 — Clarify

> "Before I sketch, can I confirm?"

- Users: 300 sales, internal → SSO / dept access control? (yes)
- Data: Confluence real-time, tickets append hourly, PDFs monthly
- Interaction: Q&A conversational + classic search — chat primary
- Latency: sales expects "Google-fast" → p95 < 3s
- Cost budget: nothing wild → prefer open source at scale
- Regulatory: internal only, can stay in our cloud


### Step 2 — Estimate

- 500k docs × 5 chunks = **2.5M chunks**
- 2.5M × 384-dim × 4 B ≈ **3.8 GB** — one Postgres box with pgvector fine
- 300 users × 20 queries/day = **6000 queries/day ≈ 0.07 QPS avg, ~1 QPS peak**
- Each query ~800 in + 250 out = **1050 tokens**
- Daily: 6.3M tokens
- At openai/gpt-oss-20b (~$0.10/1M blended): **~$0.65/day ≈ ~$20/month**


### Step 3 — Architecture

```
Confluence webhook ─┐
Support ticket DB ──┼──► Ingestion service ──► Chunk + embed ──► pgvector
PDF SharePoint     ─┘         (nightly + incremental)                │
                                                                      │
                                                                      ▼
User (SSO) ──► API Gateway ──► RAG API ──► Retriever (hybrid)
                                    │                │
                                    │                ▼
                                    │          Reranker (bge)
                                    │                │
                                    │                ▼
                                    │        openai/gpt-oss-20b via Together
                                    │                │
                                    │                ▼
                                    └── answer + citations back

Cross-cutting: SSO auth, Langfuse, usage log, semantic cache
```


### Step 4 — Deep dive (interviewer asks about ingestion)

> **Three source types, three loaders — LoaderRegistry (Section 6 Day 2) that dispatches by source type.**
>
> - Confluence: REST API + webhook on page update. Delete old chunks by `page_id` metadata, re-embed, insert. Incremental.
> - Support tickets: hourly cron reads new tickets by `created_at`, chunks small (one ticket = one chunk), embeds.
> - PDFs: monthly full re-embed (low volume).
>
> Chunking: recursive 500/50. Metadata per chunk: source, source_id, department, timestamp.
> Batching: `.encode(batch_size=64)` on GPU box for monthly PDF sweep.
> Dedupe: hash every chunk text; skip identical.

Interviewer: *"What if a page is deleted?"*

> **Soft-delete flag in metadata.** Retrieval filters `deleted != true`. Actual purge nightly. Handles undo cleanly.


### Step 5 — Close

**Tradeoffs**: *"pgvector because we already run Postgres; 3.8 GB fits. At 50M chunks I'd move to Pinecone. Together's openai/gpt-oss-20b over hosted GPT-4o because it's ~10x cheaper and quality is more than enough for this task."*

**Monitoring**: *"Langfuse trace per RAG call with sub-spans. Alerts on p95 > 4s and daily spend > $30. User thumbs feed a dashboard. Weekly manual eval on a 30-question golden set."*

**Cost**: *"Dominant cost is generation. Add semantic cache first (25%+ hit rate for a 300-person team asking similar questions). Then prompt-cache the ~2k-token static system prompt. Both cut cost significantly with no quality change."*

**End with:** *"any part you want me to go deeper into?"*
