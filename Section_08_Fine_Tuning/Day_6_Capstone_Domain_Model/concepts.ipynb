## Common pitfalls (learn from others' pain)

- **You forgot to convert to the chat template.** Model learns garbage; loss looks fine.
- **Training set contains eval examples.** Eval accuracy is misleading. Split *before* any augmentation.
- **1-line examples only.** Model can't handle real-world multi-paragraph inputs.
- **Rank too low.** Underfits — try r=32 or 64.
- **Rank too high, few examples.** Overfits — try r=8 or add regularization.
- **You didn't compare to base.** Every fine-tune LOOKS good until you compare it to just prompting the base model.
