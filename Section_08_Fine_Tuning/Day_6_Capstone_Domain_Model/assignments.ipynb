# Day 6 — Capstone Deliverables


## Deliverable 1 — Data card

Write a paragraph on your chosen dataset: source, size, license, why you picked it.


## Deliverable 2 — Training run in W&B

Share the W&B URL. Train + eval curves must be visible.


## Deliverable 3 — Eval table

20 held-out prompts. For each row: expected, base model output, fine-tuned model output. Bottom row: one accuracy number for each column.


## Deliverable 4 — Ship it

Push the merged model to Hugging Face. Serve locally with Ollama. Share both URLs / commands in a short README.


## Stretch — Portfolio write-up

A 300-word blog post: what you fine-tuned, what surprised you, base vs fine-tune numbers, and one thing you'd try next.
