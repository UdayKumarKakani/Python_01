# Different Forms of Arguments

# 1. Default Arguments

We can provide a default value to an argument by using the assignment operator (=). 

````python
def greet(name, msg="Good Morning"):
    """
    This function greets to person with the provided message
    if message is not provided, it defaults to "Good Morning"
    """
    print("Hello {0} , {1}".format(name, msg))

greet("satish", "Good Night")
````

````python
#with out msg argument
greet("satish")
````

Once we have a default argument, all the arguments to its right must also have default values.

# 2. Keyword Arguments

kwargs allows you to pass keyworded variable length of arguments to a function. You should use **kwargs if you want to handle named arguments in a function

# Example:

````python
def greet(**kwargs):
    """
    This function greets to person with the provided message
    """
    if kwargs:
        print("Hello {0} , {1}".format(kwargs['name'], kwargs['msg']))
greet(name="satish", msg="Good Morning")
````

# 3. Arbitary Arguments

Sometimes, we do not know in advance the number of arguments that will be passed into a function.Python allows us to handle this kind of situation through function calls with arbitrary number of arguments.

# Example:

````python
def greet(*names):
    """
    This function greets all persons in the names tuple 
    """
    print(names)
    
    for name in names:
        print("Hello,  {0} ".format(name))

greet("satish", "murali", "naveen", "srikanth")
````

## `*args` vs `**kwargs` — Quick Reference

| Feature | `*args` | `**kwargs` |
|---------|---------|-----------|
| What it collects | Extra **positional** arguments | Extra **keyword** arguments |
| Type inside function | `tuple` | `dict` |
| Naming | `*args` is convention; `*anything` works | `**kwargs` is convention; `**anything` works |
| Example call | `f(1, 2, 3)` | `f(name="Alice", age=30)` |

```python
def demo(*args, **kwargs):
    print(args)     # tuple of positional extras
    print(kwargs)   # dict of keyword extras

demo(1, 2, 3, name="Alice", age=30)
# (1, 2, 3)
# {'name': 'Alice', 'age': 30}
```

---
## Argument Order in a Function Signature

Python requires a specific order when mixing argument types:

```
def f(positional, default=value, *args, **kwargs):
```

| Position | Type | Required? |
|----------|------|-----------|
| 1 | Positional (required) | Yes if no default |
| 2 | Default-valued | Optional |
| 3 | `*args` (variable positional) | Optional |
| 4 | `**kwargs` (variable keyword) | Optional |

> Once you set a default value, every parameter to its right must also have a default (or be `*args` / `**kwargs`).
