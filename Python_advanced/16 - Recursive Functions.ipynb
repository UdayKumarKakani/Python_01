# Recurison

We know that in Python, a function can call other functions. It is even possible for the function to call itself. These type of construct are termed as recursive functions.

# Example:

````python
#python program to print factorial of a number using recurion

def factorial(num):
    """
    This is a recursive function to find the factorial of a given number
    """
    return 1 if num == 1 else (num * factorial(num-1))

num = 5
print "Factorial of {0} is {1} ".format(num, factorial(num))
````

# Advantages

1. Recursive functions make the code look clean and elegant.

2. A complex task can be broken down into simpler sub-problems using recursion.

3. Sequence generation is easier with recursion than using some nested iteration.


# Disadvantages

1. Sometimes the logic behind recursion is hard to follow through.

2. Recursive calls are expensive (inefficient) as they take up a lot of memory and time.

3. Recursive functions are hard to debug.


# Python program to display the fibonacci sequence up to n-th term using recursive function

````python
def fibonacci(num):
    """
    Recursive function to print fibonacci sequence
    """
    return num if num <= 1 else fibonacci(num-1) + fibonacci(num-2)

nterms = 10
print("Fibonacci sequence")
for num in range(nterms):
    print(fibonacci(num))
````

---
## Anatomy of a Recursive Function

Every recursive function needs **two** things:

| Part | Purpose | Example (factorial) |
|------|---------|---------------------|
| **Base case** | The stopping condition; returns directly | `if n == 1: return 1` |
| **Recursive case** | Calls itself with a smaller problem | `return n * factorial(n-1)` |

> **Without a base case → infinite recursion → `RecursionError`.**

---
## Recursion vs Iteration

| Aspect | Recursion | Iteration (loops) |
|--------|-----------|-------------------|
| Readability | Cleaner for tree/divide-conquer problems | Cleaner for linear walks |
| Memory | Each call uses a stack frame | Constant memory |
| Speed | Usually slower (function call overhead) | Usually faster |
| Risk | `RecursionError` if too deep (~1000 in Python) | Safe for any size |
| Best for | Factorials, Fibonacci, tree traversal, DFS | Counting, summing, simple loops |

> Python's default recursion limit is ~1000. Check with `sys.getrecursionlimit()`.
