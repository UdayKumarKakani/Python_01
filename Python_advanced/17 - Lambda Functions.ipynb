# Anonymous / Lambda Function

In Python, anonymous function is a function that is defined without a name.

While normal functions are defined using the def keyword, in Python anonymous functions are defined using the lambda keyword.

Lambda functions are used extensively along with built-in functions like filter(), map()

syntax:
    
    lambda arguments: expression

# Example:

````python
double = lambda x: x*2

print(double(5))
````

````python
def double(x):
    return x * 2

print(double(5))
````

````python
#Example use with filter()
lst = [1, 2, 3, 4, 5]
even_lst = list(filter(lambda x: (x%2 == 0), lst))
print(even_lst)
````

````python
#Example use with map()
lst = [1, 2, 3, 4, 5]
new_lst = list(map(lambda x: x ** 2, lst))
print(new_lst)
````

````python
#Example use with reduce()
from functools import reduce

lst = [1, 2, 3, 4, 5]
product_lst = reduce(lambda x, y: x*y, lst)
print(product_lst)
````

---
## `lambda` vs `def` — When to Use Which

| Feature | `lambda` | `def` |
|---------|----------|-------|
| Name | Anonymous (or assigned via `=`) | Has a real name |
| Body | **Single expression** only | Multiple statements |
| Return | Implicit (the expression) | Needs explicit `return` |
| Docstring | Not supported | Supported |
| Typical use | Short, throwaway functions passed to `map`/`filter`/`sorted` | Anything else |

```python
# Equivalent definitions:
square = lambda x: x * x

def square(x):
    return x * x
```

> Use `lambda` only when the function is short and used once. For anything more, prefer `def`.

---
## Common Lambda Patterns

| Pattern | Example | Result |
|---------|---------|--------|
| Transform list | `list(map(lambda x: x*2, [1,2,3]))` | `[2, 4, 6]` |
| Filter list | `list(filter(lambda x: x>0, [-1,2,-3,4]))` | `[2, 4]` |
| Reduce list | `reduce(lambda a,b: a+b, [1,2,3])` | `6` |
| Sort by key | `sorted(words, key=lambda w: len(w))` | sorts by length |
| Sort by 2nd field | `sorted(pairs, key=lambda p: p[1])` | sorts by 2nd item |
| Conditional value | `(lambda x: "even" if x%2==0 else "odd")(4)` | `"even"` |
