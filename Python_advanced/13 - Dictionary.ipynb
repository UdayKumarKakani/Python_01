# Dictionary

Python dictionary is an unordered collection of items. While other compound data types have only value as an element, a dictionary has a key: value pair.

# Dict Creation

````python
#empty dictionary
my_dict = {}

#dictionary with integer keys
my_dict = {1: 'abc', 2: 'xyz'}
print(my_dict)

#dictionary with mixed keys
my_dict = {'name': 'satish', 1: ['abc', 'xyz']}
print(my_dict)


#create empty dictionary using dict()
my_dict = dict()

my_dict = dict([(1, 'abc'), (2, 'xyz')])    #create a dict with list of tuples
print(my_dict)
````

# Dict Access

````python
my_dict = {'name': 'satish', 'age': 27, 'address': 'guntur'}

#get name
print(my_dict['name'])
````

````python
#if key is not present it gives KeyError
print(my_dict['degree'])
````

````python
#another way of accessing key
print(my_dict.get('address'))
````

````python
#if key is not present it will give None using get method
print(my_dict.get('degree'))
````

# Dict Add or Modify  Elements

````python
my_dict = {'name': 'satish', 'age': 27, 'address': 'guntur'}

#update name 
my_dict['name'] = 'raju'

print(my_dict)
````

````python
#add new key
my_dict['degree'] = 'M.Tech'

print(my_dict)
````

# Dict Delete or Remove Element

````python
#create a dictionary
my_dict = {'name': 'satish', 'age': 27, 'address': 'guntur'}

#remove a particular item
print(my_dict.pop('age'))

print(my_dict)
````

````python
my_dict = {'name': 'satish', 'age': 27, 'address': 'guntur'}

#remove an arbitarty key
my_dict.popitem()

print(my_dict)
````

````python
squares = {2: 4, 3: 9, 4: 16, 5: 25}

#delete particular key
del squares[2]

print(squares)
````

````python
#remove all items
squares.clear()

print(squares)
````

````python
squares = {2: 4, 3: 9, 4: 16, 5: 25}

#delete dictionary itself
del squares

print(squares) #NameError because dict is deleted
````

# Dictionary Methods

````python
squares = {2: 4, 3: 9, 4: 16, 5: 25}

my_dict = squares.copy()
print(my_dict)
````

````python
#fromkeys[seq[, v]] -> Return a new dictionary with keys from seq and value equal to v (defaults to None).
subjects = {}.fromkeys(['Math', 'English', 'Hindi'], 0)
print(subjects)
````

````python
subjects = {2:4, 3:9, 4:16, 5:25}
print(subjects.items()) #return a new view of the dictionary items (key, value)
````

````python
subjects = {2:4, 3:9, 4:16, 5:25}
print(subjects.keys()) #return a new view of the dictionary keys
````

````python
subjects = {2:4, 3:9, 4:16, 5:25}
print(subjects.values()) #return a new view of the dictionary values
````

````python
#get list of all available methods and attributes of dictionary
d = {}
print(dir(d))
````

# Dict Comprehension

````python
#Dict comprehensions are just like list comprehensions but for dictionaries

d = {'a': 1, 'b': 2, 'c': 3}
for pair in d.items():
    print(pair)
````

````python
#Creating a new dictionary with only pairs where the value is larger than 2
d = {'a': 1, 'b': 2, 'c': 3, 'd': 4}
new_dict = {k:v for k, v in d.items() if v > 2}
print(new_dict)
````

````python
#We can also perform operations on the key value pairs
d = {'a':1,'b':2,'c':3,'d':4,'e':5}
d = {k + 'c':v * 2 for k, v in d.items() if v > 2}
print(d)
````

## Common Dictionary Methods Reference

| Method | Description | Example |
|--------|-------------|---------|
| `d[key]` | Get value (raises `KeyError` if missing) | `d["name"]` |
| `d.get(key, default)` | Get value (returns `default` if missing) | `d.get("age", 0)` |
| `d[key] = value` | Add or update an entry | `d["age"] = 30` |
| `d.pop(key)` | Remove key and return its value | `d.pop("age")` |
| `d.popitem()` | Remove and return last (key, value) | `d.popitem()` |
| `d.update(other)` | Merge another dict into this one | `d.update({"x": 1})` |
| `d.keys()` | View of all keys | `list(d.keys())` |
| `d.values()` | View of all values | `list(d.values())` |
| `d.items()` | View of all (key, value) pairs | `for k,v in d.items():` |
| `d.setdefault(k, v)` | Get value; insert `v` if key missing | `d.setdefault("x", [])` |
| `d.clear()` | Remove all items | `d.clear()` |
| `key in d` | Check if key exists | `"name" in d` |

## Looping through Dictionaries — the Right Way

```python
prices = {"apple": 50, "banana": 20, "mango": 80}

# Loop over keys (default)
for fruit in prices:
    print(fruit)

# Loop over values
for price in prices.values():
    print(price)

# Loop over key-value pairs (most common!)
for fruit, price in prices.items():
    print(f"{fruit}: Rs.{price}")
```

````python
prices = {"apple": 50, "banana": 20, "mango": 80}

for fruit, price in prices.items():
    print(f"{fruit}: Rs.{price}")
````
