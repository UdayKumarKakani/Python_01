# Tuples

-> A tuple is similar to list

-> The diffence between the two is that we can't change the elements of tuple once it is assigned whereas in the list, elements can be changed

# Tuple creation

````python
#empty tuple
t = ()

#tuple having integers
t = (1, 2, 3)
print(t)

#tuple with mixed datatypes
t = (1, 'raju', 28, 'abc')
print(t)

#nested tuple
t = (1, (2, 3, 4), [1, 'raju', 28, 'abc'])
print(t)
````

````python
#only parenthesis is not enough
t = ('satish')
type(t)
````

````python
#need a comma at the end
t = ('satish',)
type(t)
````

````python
#parenthesis is optional
t = "satish", 
print(type(t))

print(t)
````

# Accessing Elements in Tuple

````python
t = ('satish', 'murali', 'naveen', 'srinu', 'brahma')

print(t[1])
````

````python
#negative index
print(t[-1]) #print last element in a tuple
````

````python
#nested tuple
t = ('ABC', ('satish', 'naveen', 'srinu'))

print(t[1])
````

````python
print(t[1][2])
````

````python
#Slicing
t = (1, 2, 3, 4, 5, 6)

print(t[1:4])

#print elements from starting to 2nd last elements
print(t[:-2])

#print elements from starting to end
print(t[:])
````

# Changing a Tuple

#unlike lists, tuples are immutable
#This means that elements of a tuple cannot be changed once it has been assigned. But, if the element is itself a mutable datatype like list, its nested items can be changed.

````python
#creating tuple
t = (1, 2, 3, 4, [5, 6, 7])

t[2] = 'x' #will get TypeError
````

````python
t[4][1] = 'satish'
print(t)
````

````python
#concatinating tuples

t = (1, 2, 3) + (4, 5, 6)
print(t)
````

````python
#repeat the elements in a tuple for a given number of times using the * operator.
t = (('satish', ) * 4)
print(t)
````

# Tuple Deletion

## Tuple vs List — Quick Comparison

| Feature | Tuple `()` | List `[]` |
|---------|-----------|-----------|
| Mutability | **Immutable** (can't change) | **Mutable** (can change) |
| Syntax | `(1, 2, 3)` | `[1, 2, 3]` |
| Performance | Slightly faster | A bit slower |
| Use as dict key | ✅ Yes (hashable) | ❌ No |
| Methods | Only `count()`, `index()` | Many (`append`, `pop`, etc.) |
| Best for | Fixed data, records | Collections that grow/change |

## Tuple Unpacking — Powerful Feature

Tuples can be **unpacked** into individual variables in one line:

```python
# Basic unpacking
point = (3, 4)
x, y = point
print(x, y)   # 3 4

# Swap variables (no temp variable needed!)
a, b = 1, 2
a, b = b, a
print(a, b)   # 2 1

# Star unpacking — collect remaining elements
first, *rest = (1, 2, 3, 4, 5)
print(first, rest)   # 1 [2, 3, 4, 5]
```

````python
# Tuple unpacking demo
point = (10, 20)
x, y = point
print("x =", x, "y =", y)

# Swap
a, b = 5, 9
a, b = b, a
print("After swap:", a, b)

# Star unpacking
first, *middle, last = (1, 2, 3, 4, 5)
print(first, middle, last)
````
