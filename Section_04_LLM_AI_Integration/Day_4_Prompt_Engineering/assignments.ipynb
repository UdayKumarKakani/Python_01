```
import os
from dotenv import load_dotenv
from together import Together

load_dotenv()
client = Together()
MODEL = "meta-llama/Meta-Llama-3.1-8B-Instruct-Turbo"

def ask(prompt, system=None, max_tokens=300, temperature=0.2):
    messages = []
    if system: messages.append({"role":"system","content":system})
    messages.append({"role":"user","content":prompt})
    r = client.chat.completions.create(model=MODEL, messages=messages,
                                        max_tokens=max_tokens, temperature=temperature)
    return r.choices[0].message.content.strip()
```

## Exercise 2 — Few-shot email classifier

Using the few-shot pattern, classify these three new messages as `SUPPORT`, `SALES`, or `SPAM`:

1. *"Password reset link isn't arriving in my inbox."*
2. *"What's the price for 100 seats?"*
3. *"URGENT!!! You won a free iPhone — reply YES."*

Expected: `SUPPORT`, `SALES`, `SPAM`.

```
TEST = [
    "Password reset link isn't arriving in my inbox.",
    "What's the price for 100 seats?",
    "URGENT!!! You won a free iPhone — reply YES.",
]

def classify(msg):
    prompt = f'''Classify each customer message as SUPPORT, SALES, or SPAM.

Message: "My login isn't working since the update."
Label: SUPPORT

Message: "Do you offer discounts for annual subscriptions?"
Label: SALES

Message: "CLICK HERE FOR FREE CRYPTO!!!"
Label: SPAM

Message: "{msg}"
Label:'''
    return ask(prompt, max_tokens=5)

for msg in TEST:
    print(f"{classify(msg):8s} :: {msg}")
```
