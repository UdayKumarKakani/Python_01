## Exercise 3 — Track your token bill

Call `gpt-4o-mini` with any short prompt. Print:

- Input tokens (`resp.usage.prompt_tokens`)
- Output tokens (`resp.usage.completion_tokens`)
- Estimated cost in dollars

Use these prices:
- Input: **$0.15 / million tokens**
- Output: **$0.60 / million tokens**

💡 **Formula:** `cost = (in/1_000_000)*0.15 + (out/1_000_000)*0.60`

```
resp = oai.chat.completions.create(
    model="gpt-4o-mini",
    messages=[{"role":"user","content":"Give me 3 fun facts about octopuses."}],
)

in_tokens  = resp.usage.prompt_tokens
out_tokens = resp.usage.completion_tokens

# TODO: compute cost using the prices above
cost = 0.0

print("reply:\n", resp.choices[0].message.content, "\n")
print(f"input tokens : {in_tokens}")
print(f"output tokens: {out_tokens}")
print(f"cost         : ${cost:.6f}")
```
