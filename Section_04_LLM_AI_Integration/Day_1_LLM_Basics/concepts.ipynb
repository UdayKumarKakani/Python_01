## 1. An LLM is very fancy autocomplete

Think of the autocomplete on your phone. You type *"I'll be a little bit"* and it suggests *"late"*.

An LLM does exactly the same thing — just at a much bigger scale.

- **Your phone** predicts the next word from the last few words.
- **An LLM** (like ChatGPT or Claude) predicts the next word from *everything* you've written (a whole essay, a whole conversation).

That's it. The magic is that when a system this good at predicting the next word sees enough of the internet, it starts to *look* like it understands you.

## 2. Tokens: how AI chops up words

An LLM doesn't see letters. It doesn't quite see words either. It sees **tokens** — little chunks somewhere in between.

| Text | Tokens (chunks it sees) |
|---|---|
| `"cat"` | `["cat"]` — 1 token |
| `"unbelievable"` | `["un", "believ", "able"]` — 3 tokens |
| `"🎉"` | usually 2–3 tokens |

**Rule of thumb:** 1 token ≈ 4 English characters ≈ ¾ of a word.

A page of text? Roughly **500 tokens**.

**Why you care:** every AI provider charges you *per token*. More tokens = more money. We'll see the price tag in Day 2.

```python
# Optional peek — how many tokens is a sentence?
# Skip this cell if you're not a coder — the number matters, not the code.
!pip install tiktoken --quiet
```

```python
import tiktoken

enc = tiktoken.encoding_for_model("gpt-4o-mini")

sentences = [
    "Hello, world! ",
    "The quick brown fox jumps over the lazy dog.",
    "Unbelievable — this is a longer sentence with several words.",
]

for s in sentences:
    n = len(enc.encode(s))
    print(f"{n:3d} tokens :  {s}")
```

**What just happened:** `tiktoken` is a tool that shows you exactly how OpenAI counts tokens. The longer/rarer the words, the more tokens it costs.

## 3. The (very short) tour of what's inside

Inside an LLM, three things happen every time it generates a word:

1. **Tokenize** — your text is chopped into tokens (like we just saw).
2. **Attention** — the model looks back at *every* previous token and decides which ones matter most for guessing the next one.
3. **Predict** — it picks the next most likely token, adds it to the response, and repeats.

### Attention — the one big idea

Read this sentence:

> *"The animal didn't cross the street because it was too tired."*

What does **"it"** refer to? Obviously *the animal*. You know that because you paid attention to the word "animal" earlier in the sentence.

The LLM does the exact same thing — it looks back at every previous word and weights how much attention to pay to each one. That mechanism, called **self-attention**, is what made ChatGPT possible.

## 4. "Temperature" — the creativity dial

Every time an LLM picks the next word, it actually has a list of possible next words with probabilities.

- **Temperature 0** → *always* pick the most likely word. Boring, deterministic, factual.
- **Temperature 1** → pick according to the natural probabilities. Balanced.
- **Temperature 1.5+** → pick unusual words on purpose. Creative, sometimes gibberish.

**Rule of thumb:**

| Task | Temperature |
|---|---|
| Extract structured data | 0 |
| Customer support answer | 0.2 – 0.4 |
| Brainstorming | 0.8 – 1.0 |
| Poetry / creative writing | 1.0 – 1.3 |

## 5. Why do LLMs make things up? ("Hallucinations")

Because they're autocomplete.

An LLM doesn't have a database of facts. It doesn't know what it knows. It's trained to write *plausible-looking* next words. If it doesn't know the answer, it writes a plausible-*sounding* wrong answer.

**Real example:** ask ChatGPT about a fake court case ("Smith v. Johnson 2019"). It will happily *invent* a summary — a lawyer in New York actually filed hallucinated citations in court in 2023.

**How to reduce hallucinations** (previewing later days):
- Use lower temperature.
- Feed the model the actual documents (Section 5 — RAG).
- Ask it to say "I don't know" when unsure.
- Verify structured outputs (Day 4).

## Recap

- An LLM = **very good autocomplete**.
- Text is chopped into **tokens** (~¾ of a word each). You pay per token.
- **Attention** lets the model look at any earlier word when picking the next one.
- **Temperature** controls how random / creative the output is.
- **Hallucinations** happen because the model optimizes for plausible, not true.
- Different models have different strengths — the job of an AI engineer is to route work to the right one.
