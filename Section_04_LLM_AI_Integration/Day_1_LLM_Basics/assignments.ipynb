## Exercise 1 — Count tokens in your own sentence

Write any sentence you like and print how many tokens it uses. Then guess: which of the three sentences below costs the most?

1. `"Hi."`
2. `"The seven wonders of the ancient world."`
3. `"🎉🚀🎊 Party emojis are surprisingly expensive!"`

💡 **Tip:** the emoji one usually wins.

```
import tiktoken
enc = tiktoken.encoding_for_model("gpt-4o-mini")

sentences = [
    "Hi.",
    "The seven wonders of the ancient world.",
    "🎉🚀🎊 Party emojis are surprisingly expensive!",
    # TODO: add your own sentence here
]

for s in sentences:
    print(f"{len(enc.encode(s)):3d} tokens : {s}")
```

## Exercise 2 — Estimate the cost of a page

Below is a chunk of text roughly one page long. Count the tokens, then work out roughly what it would cost to send it to `gpt-4o-mini`.

OpenAI charges about **$0.15 per 1 million input tokens** for `gpt-4o-mini`.

💡 **Formula:** cost = (tokens ÷ 1,000,000) × 0.15

```
page = ("Lorem ipsum dolor sit amet, consectetur adipiscing elit. " * 60)

tokens = len(enc.encode(page))
cost = (tokens / 1_000_000) * 0.15

print(f"tokens : {tokens}")
print(f"cost   : ${cost:.6f}")
print(f"To send this page 1,000 times: ${cost*1000:.4f}")
```

## Exercise 3 — Feel the temperature dial

Below is a fake probability distribution for the next word. Change `T` from `0.1` to `1.5` and re-run to see how the distribution shifts.

- `T=0.1` — one word dominates (deterministic)
- `T=1.0` — natural distribution
- `T=1.5` — flatter, more random

💡 **Tip:** you don't need to understand the math — just watch the numbers move.

```
import numpy as np

words = ["delicious", "tasty", "good", "ok", "awful"]
logits = np.array([2.0, 1.5, 1.0, 0.2, -1.0])

T = 1.0  # try 0.1 and 1.5 too

z = logits / T
e = np.exp(z - z.max())
probs = e / e.sum()

for w, p in zip(words, probs):
    print(f"{w:12s} {p:.2%}")
```
