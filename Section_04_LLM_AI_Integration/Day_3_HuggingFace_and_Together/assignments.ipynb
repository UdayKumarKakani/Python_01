## Exercise 2 — Zero-shot classification (no training required!)

`pipeline("zero-shot-classification")` can classify text into *any labels you invent*, without training. Try it with 3 labels of your choice:

```python
labels = ["food", "sports", "technology"]
```

Classify: `"The cake was moist and delicious."`

💡 **Expected:** highest score on `food`.

````
clf = pipeline("zero-shot-classification", model="facebook/bart-large-mnli")

result = clf(
    "The cake was moist and delicious.",
    candidate_labels=["food", "sports", "technology"],
)

for label, score in zip(result["labels"], result["scores"]):
    print(f"  {label:12s} {score:.2%}")
````
